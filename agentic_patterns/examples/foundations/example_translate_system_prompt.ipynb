# Simple LLM: Translation (System Prompt)

```python
import rich
from agentic_patterns.core.agents import get_agent, run_agent
```

```python
language = "French"

# System prompt
system_prompt = f"Translate into {language}"
agent = get_agent(system_prompt=system_prompt)

# User prompt
prompt = "I like coffee."

# Run agent with verbose output
agent_run, nodes = await run_agent(agent, prompt, verbose=True)

assert agent_run is not None and agent_run.result is not None
print(agent_run.result.output)
```
