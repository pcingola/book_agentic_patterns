# Adversarial & Debate Agents

This notebook demonstrates two reusable adversarial orchestration patterns:
structured debate (advocate vs. critic with arbiter) and red-team analysis.

## Imports

```python
from agentic_patterns.core.agents import DebateOrchestrator, RedTeamAgent
```

## Define a proposal

```python
proposal = "Our e-commerce platform should migrate its primary database from PostgreSQL to MongoDB to better handle the growing variety of product catalog schemas."
```

## Basic debate

```python
debate = DebateOrchestrator(max_rounds=2)
result = await debate.run(proposal)
```

```python
for i, rnd in enumerate(result.rounds):
    print(f"--- Round {i + 1} ---")
    print(f"Advocate ({rnd.advocate.position}):")
    for arg in rnd.advocate.arguments:
        print(f"  - {arg.claim}")
    print(f"Critic ({rnd.critic.position}):")
    for arg in rnd.critic.arguments:
        print(f"  - {arg.claim}")
    print()
```

```python
print(f"Decision: {result.verdict.decision}")
print(f"Reasoning: {result.verdict.reasoning}")
print(f"Accepted: {result.verdict.accepted_claims}")
print(f"Rejected: {result.verdict.rejected_claims}")
print(f"Open questions: {result.verdict.open_questions}")
```

## Persona-driven debate

Pass role descriptions as `advocate_prompt` / `critic_prompt` to simulate different viewpoints.

```python
debate_persona = DebateOrchestrator(
    advocate_prompt="You are a startup CTO who values developer velocity and schema flexibility above all else.",
    critic_prompt="You are a database reliability engineer who has managed PostgreSQL clusters at scale for 10 years.",
    max_rounds=2,
)
result_persona = await debate_persona.run(proposal)
```

```python
print(f"Decision: {result_persona.verdict.decision}")
print(f"Reasoning: {result_persona.verdict.reasoning}")
print(f"Open questions: {result_persona.verdict.open_questions}")
```

## Red-team analysis

Apply red-team analysis to the debate verdict to surface remaining risks.

```python
red_team = RedTeamAgent(
    threat_model="Database migration risks: data loss, downtime, performance regression, team skill gaps, vendor lock-in."
)
rt_result = await red_team.analyze(
    result=result.verdict.decision,
    context=result.verdict.reasoning,
)
```

```python
print(f"Summary: {rt_result.summary}\n")
for ch in rt_result.challenges:
    print(f"[{ch.severity}] {ch.claim}")
    print(f"  Attack: {ch.attack}")
    print(f"  Required evidence: {ch.required_evidence}")
    print()
```
